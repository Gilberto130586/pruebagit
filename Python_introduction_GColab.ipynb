###how can we change the header columns names to caps

```python
import csv
```

```python
source_path = "/content/sample_data/california_housing_test.csv"
target_path = "/content/sample_data/california_housing_test_updated.csv"

with open(source_path, 'r') as infile, open(target_path, 'w', newline='') as outfile:
    reader = csv.reader(infile)
    writer = csv.writer(outfile)

    header = next(reader)  # Read the header row
    header = [col.upper() for col in header]  # Convert header to uppercase
    writer.writerow(header)  # Write the updated header

    for row in reader:
        writer.writerow(row)  # Write the rest of the rows
```

###How can we change the datatype into some columns

```python
# prompt: into california_housing_test_updated.csv file, Change housing_median_age to integer

import csv

source_path = "/content/sample_data/california_housing_test_updated.csv"
target_path = "/content/sample_data/california_housing_test_updated2.csv"

with open(source_path, 'r') as infile, open(target_path, 'w', newline='') as outfile:
    reader = csv.DictReader(infile)
    writer = csv.DictWriter(outfile, fieldnames=reader.fieldnames)
    writer.writeheader()

    for row in reader:
        row['HOUSING_MEDIAN_AGE'] = int(float(row['HOUSING_MEDIAN_AGE']))  # Convert to integer
        writer.writerow(row)
```

**Add new column**

```python
# prompt: into california_housing_test_updated2.csv, add new column "median_age_house_value_ratio" named

source_path = "/content/sample_data/california_housing_test_updated2.csv"
target_path = "/content/sample_data/california_housing_test_updated3.csv"

with open(source_path, 'r') as infile, open(target_path, 'w', newline='') as outfile:
    reader = csv.DictReader(infile)
    fieldnames = reader.fieldnames + ["MEDIAN_AGE_HOUSE_VALUE_RATIO"]  # Add the new column name
    writer = csv.DictWriter(outfile, fieldnames=fieldnames)
    writer.writeheader()

    for row in reader:
        # Calculate the new column value (replace with your actual calculation)
        if float(row['MEDIAN_HOUSE_VALUE']) > 0:
            ratio = float(row['HOUSING_MEDIAN_AGE']) / float(row['MEDIAN_HOUSE_VALUE'])
        else:
            ratio = 0  # Handle cases where MEDIAN_HOUSE_VALUE is zero
        row["MEDIAN_AGE_HOUSE_VALUE_RATIO"] = ratio
        writer.writerow(row)
```
